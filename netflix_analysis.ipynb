```python
# Convert date column
df['Release_Date'] = pd.to_datetime(df['Release_Date'])

# Convert numeric columns
df['Vote_Count'] = pd.to_numeric(df['Vote_Count'])
df['Vote_Average'] = pd.to_numeric(df['Vote_Average'])

# Verify
df.info()
```

```python
import matplotlib.pyplot as plt

df['Genre'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Genres')
plt.show()
```

```python
df['Original_Language'].value_counts().head(10)
```

```python
df[['Title','Vote_Average']].sort_values(
    by='Vote_Average',
    ascending=False
).head(10)
```

```python
df[['Title','Popularity']].sort_values(
    by='Popularity',
    ascending=False
).head(10)
```

```python
df['Year'] = df['Release_Date'].dt.year
```

```python
df['Year'].value_counts().sort_index().plot(figsize=(12,5))
plt.title('Movies Released Per Year')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df[['Popularity','Vote_Count','Vote_Average']].corr()

plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True)
plt.title('Correlation Matrix')
plt.show()
```
